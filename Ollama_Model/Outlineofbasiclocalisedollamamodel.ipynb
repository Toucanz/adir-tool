```python
with open("interviewtranscript.txt","r") as file:
    transcript = file.read()


print(transcript)
```

```python
system_prompt = "You are a assistant that anaylses interview transcripts and writes a summary. Here is the expected format:" \
"Example Transcript:" \
"Speaker 1: When the subject was 4 to 5 years of age, did they look at you directly in the face when doing things with you or talking with you? Could you catch their eye? Did they sometimes watch you as you walked into the room? Would they have looked back and forth to your face as other children would? What about with others?" \
"Speaker 2: At that age, he did look at me sometimes, but not often. If we were playing together, I could usually catch his eye for a second or two, but then he’d look away. When I walked into the room, he didn’t really watch me like other kids might. He seemed more focused on whatever he was doing. With others, it was about the same, he didn’t seek out eye contact much, even with family members." \
"Expected Output:" \
"When Brian was between 4 and 5 years old, he displayed inconsistent eye contact in social interactions. His mother noted that his gaze was fleeting and he rarely watched her like other kids might have. Name seemed very focussed on his concrete task and did not seek out eye contact with other family members."
```

```python
def user_prompt_for(transcript):
    user_prompt = (
        "Here is a new interview transcript. Please summarize it in the same style\n"
        "Please summarize it in the style above"
        
    )
    user_prompt += transcript
    return user_prompt
```

```python
print(user_prompt_for(transcript))
```

```python
user_prompt = user_prompt_for(transcript)

messages = [
        {"role": "system", "content": system_prompt},
        {"role":"user", "content": user_prompt}
    ]
```

```python
#This is loading the payload
payload = {
    "model": MODEL,
    "messages": messages
    
}
```

```python
#Getting the response
import ollama

response = ollama.chat(model=MODEL, messages=messages)

```
