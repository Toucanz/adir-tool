```python
with open("interviewtranscript.txt","r") as file:
    transcript = file.read()


print(transcript)
```

```python
system_prompt = "You are a psychological assistant that anaylses the content of transcripts and provides a clinical summary. Formulate the summary appropriate to a clinical report"
```

```python
def user_prompt_for(transcript):
    user_prompt = (
        "You are reading an interview transcript between Speaker 1 and Speaker 2.\n"
        "This interview transcript is a diagnostic interview for Autism of the child of Speaker 2.\n"
        "Focus on clinically relevant data.\n\n"
    )
    user_prompt += transcript
    return user_prompt
```

```python
print(user_prompt_for(transcript))
```

```python
user_prompt = user_prompt_for(transcript)

messages = [
        {"role": "system", "content": system_prompt},
        {"role":"user", "content": user_prompt}
    ]
```

```python
#This is loading the payload
payload = {
    "model": MODEL,
    "messages": messages
    
}
```

```python
#Getting the response
import ollama

response = ollama.chat(model=MODEL, messages=messages)

```
