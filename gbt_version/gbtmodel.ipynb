```python
def user_prompt_for(transcript):
    user_prompt = (
        "Here is a new interview transcript. Summarize it in the same style as given in output"
        
    )
    user_prompt += transcript
    return user_prompt
```

```python
def summarize_transcript(user_prompt):
    response = client.chat.completions.create(
        model="gpt-4o-mini", 
        messages=[
            {"role": "system", "content": "You are an assistant that summarizes interview transcripts for clinical reports."},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content
```
